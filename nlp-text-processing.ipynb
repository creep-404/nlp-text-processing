Combine all the cleaning steps from Q3 to Q7 into a single cleaning function.
Apply the complete pipeline on the original text column.
Create a new column called cleaned_text.
Display the original text and cleaned text side by side for 10 samples.

```python
import pandas as pd
import nltk
import string
import re
from nltk.corpus import stopwords

# Download required NLTK resource
nltk.download("stopwords")

# Load the dataset
df = pd.read_csv("train.txt", sep=";", names=["text", "emotions"])

# Load English stopwords
stop_words = set(stopwords.words("english"))

# Complete text cleaning function
def clean_text(text):
    # Convert to lowercase
    text = text.lower()

    # Remove punctuation
    text = text.translate(str.maketrans("", "", string.punctuation))

    # Remove digits
    text = re.sub(r"\d+", "", text)

    # Keep only ASCII characters (remove emojis and special symbols)
    text = text.encode("ascii", "ignore").decode("ascii")

    # Remove stopwords
    words = text.split()
    words = [word for word in words if word not in stop_words]

    # Join words back into a sentence
    return " ".join(words)

# Apply the cleaning function
df["cleaned_text"] = df["text"].apply(clean_text)

# Display original and cleaned text side by side
print("=" * 120)
print("ORIGINAL TEXT VS CLEANED TEXT")
print("=" * 120)

for i in range(10):
    print(f"\nSample {i+1}")
    print("Original :", df["text"].iloc[i])
    print("Cleaned  :", df["cleaned_text"].iloc[i])

# Display first 10 rows
print("\n" + "=" * 120)
print("FIRST 10 ROWS")
print("=" * 120)
print(df[["text", "cleaned_text"]].head(10))

# Dataset information
print("\nDataset Shape:", df.shape)
print("Total Rows Processed:", len(df))
```

After cleaning:
Calculate the number of words in each cleaned text.
Plot a histogram of text lengths.
Print the average, minimum, and maximum text length.
Write observations about the cleaned data.

```python
import pandas as pd
import nltk
import string
import re
import matplotlib.pyplot as plt
from nltk.corpus import stopwords

# Download required NLTK resource
nltk.download("stopwords")

# Load the dataset
df = pd.read_csv("train.txt", sep=";", names=["text", "emotions"])

# Load English stopwords
stop_words = set(stopwords.words("english"))

# Complete cleaning function
def clean_text(text):
    # Convert to lowercase
    text = text.lower()

    # Remove punctuation
    text = text.translate(str.maketrans("", "", string.punctuation))

    # Remove digits
    text = re.sub(r"\d+", "", text)

    # Keep only ASCII characters (remove emojis/special symbols)
    text = text.encode("ascii", "ignore").decode("ascii")

    # Remove stopwords
    words = text.split()
    words = [word for word in words if word not in stop_words]

    return " ".join(words)

# Apply cleaning
df["cleaned_text"] = df["text"].apply(clean_text)

# Calculate number of words in each cleaned text
df["text_length"] = df["cleaned_text"].apply(lambda x: len(x.split()))

# Display first 10 samples
print("=" * 100)
print("ORIGINAL TEXT | CLEANED TEXT | WORD COUNT")
print("=" * 100)

print(df[["text", "cleaned_text", "text_length"]].head(10))

# Plot histogram of text lengths
plt.figure(figsize=(8,5))
plt.hist(df["text_length"], bins=20, edgecolor="black")
plt.title("Histogram of Cleaned Text Lengths")
plt.xlabel("Number of Words")
plt.ylabel("Frequency")
plt.grid(True)
plt.show()

# Print statistics
print("\nText Length Statistics")
print("-" * 30)
print("Average text length :", round(df["text_length"].mean(), 2))
print("Minimum text length :", df["text_length"].min())
print("Maximum text length :", df["text_length"].max())

# Observations
print("\nObservations:")
print("1. The text has been cleaned by removing punctuation, digits, emojis, and stopwords.")
print("2. Most cleaned texts contain fewer words than the original texts.")
print("3. The histogram shows the distribution of text lengths after preprocessing.")
print("4. The cleaned data is more consistent and suitable for NLP and machine learning tasks.")
```

Using the Emotions Dataset:

1. Load the data
2. Encode the emotion labels
3. Perform complete text cleaning (lowercase → punctuation → numbers → emojis → stopwords)
4. Save the cleaned dataset as cleaned\_emotions.csv
5. Show value counts of each emotion

```python
import pandas as pd
import nltk
import string
import re
from nltk.corpus import stopwords
from sklearn.preprocessing import LabelEncoder

# Download required NLTK resource
nltk.download("stopwords")

# Load the dataset
df = pd.read_csv("train.txt", sep=";", names=["text", "emotions"])

# Encode the emotion labels
label_encoder = LabelEncoder()
df["emotion_encoded"] = label_encoder.fit_transform(df["emotions"])

# Load English stopwords
stop_words = set(stopwords.words("english"))

# Complete text cleaning function
def clean_text(text):
    # Convert to lowercase
    text = text.lower()

    # Remove punctuation
    text = text.translate(str.maketrans("", "", string.punctuation))

    # Remove numbers
    text = re.sub(r"\d+", "", text)

    # Remove emojis and special symbols (keep only ASCII)
    text = text.encode("ascii", "ignore").decode("ascii")

    # Remove stopwords
    words = text.split()
    words = [word for word in words if word not in stop_words]

    # Join words back into a sentence
    return " ".join(words)

# Apply text cleaning
df["cleaned_text"] = df["text"].apply(clean_text)

# Save the cleaned dataset
df.to_csv("cleaned_emotions.csv", index=False)

# Display the first 10 rows
print("=" * 100)
print("FIRST 10 ROWS OF CLEANED DATASET")
print("=" * 100)
print(df.head(10))

# Show emotion counts
print("\n" + "=" * 50)
print("EMOTION VALUE COUNTS")
print("=" * 50)
print(df["emotions"].value_counts())

# Display emotion-to-code mapping
print("\n" + "=" * 50)
print("EMOTION ENCODING")
print("=" * 50)

emotion_mapping = dict(zip(label_encoder.classes_,
                           label_encoder.transform(label_encoder.classes_)))

print(emotion_mapping)

# Confirmation message
print("\nCleaned dataset saved successfully as 'cleaned_emotions.csv'")
```
